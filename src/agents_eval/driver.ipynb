```python
import os

from agents_eval.agent import Agent
from agents_eval.environments import MCPEnvironment
from agents_eval.evaluation import Evaluation
from agents_eval.models import GPTJRCModel
from agents_eval.paradigms import ReflectionParadigm
```

```python
# Notebook Params
class Params:
    CPU_ONLY = False
    MULTI_GPU = True
    GPU = 0


if not Params.MULTI_GPU:
    os.environ["CUDA_VISIBLE_DEVICES"] = str(Params.GPU)
if Params.CPU_ONLY:
    os.environ["CUDA_VISIBLE_DEVICES"] = "-1"
```

```python
agent = Agent(
    # OpenAICompatibleModel(model_id="Qwen/Qwen3-8B", local=True),
    GPTJRCModel(),
    ReflectionParadigm(),
    verbose=True,
    _native_function_calling=False,
)
agent.run(
    environment=MCPEnvironment(mcp_server="local"),
    task="What are the EC numbers of the two most commonly used chemicals for the virus testing method in the paper about SPFMV and SPCSV in the Pearl Of Africa from 2016? Return the semicolon-separated numbers in the order of the alphabetized chemicals.",
)
```

```python
evaluation = Evaluation(verbose=True)
evaluation.evaluate_all(
    models=[
        GPTJRCModel(),
        # HuggingfaceModel("Qwen/Qwen3-8B"),
    ],
    paradigms=[ReflectionParadigm()],
    environments=[MCPEnvironment(mcp_server="local")],
    tasklist="GAIA",
)
```
